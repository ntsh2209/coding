```python
import pandas as pd

df = pd.read_excel('data_set/sample_pivot.xlsx', parse_dates=['Date'])

print(df.head())
```

```python
pivot = pd.pivot_table(df, index = ['Region', 'Type'], values = 'Sales', aggfunc = 'sum')

print(pivot)
```

```python
pivot.style.format({'Sales':'${0:,.0f}'})
```

```python
# Float with 2 decimal places: {:.2f}
# Pad numbers with zeroes: {:0>2d}
# Percent with 2 decimal places: {:.2%}
```

```python
def highlight_fifty(val):
    color = 'red' if val > 50000 else 'black'
    return 'color: %s' % color
```

```python
pivot_highlight = pivot.style.applymap(highlight_fifty)
pivot_highlight
```

```python
pivot_highlight = pivot.style.applymap(highlight_fifty).format({'Sales':'${0:,.0f}'})
pivot_highlight
```

```python
pivot_highlight = pivot.style.format({'Sales':'${0:,.0f}'})\
   .highlight_max(color='green')\
   .highlight_min(color='red')
pivot_highlight
```

```python
pivot.style.format({'Sales':'${0:,.0f}'}) \
   .highlight_max(color='green') \
   .highlight_min(color='red')
```

```python
pivot = pd.pivot_table(
    data=df,
    index='Region',
    values='Sales'
)
print(pivot)
```

```python
pivot = pd.pivot_table(
    data=df,
    index='Region',
    values=['Sales','Units'],
    aggfunc='sum'
)
print(pivot)
```

```python
pivot = pd.pivot_table(
    data=df,
    index='Region',
    columns='Type',
    values='Sales'
)
print(pivot)
```

```python
pivot = pd.pivot_table(
    data=df,
    index=['Region',df['Date'].dt.quarter],
    columns='Type',
    values='Sales'
)

print(pivot.head())
```

```python
print(pivot.loc[('East', 1), "Men's Clothing"])
```

```python
# Adding totals to rows and columns
pivot = pd.pivot_table(
    data=df,
    index='Region',
    columns='Type',
    values='Sales',
    margins=True
)
print(pivot)
```

```python
# Adding totals to rows and columns
pivot = pd.pivot_table(
    data=df,
    index='Region',
    columns='Type',
    values='Sales',
    margins=True,
    margins_name='Total'
)
print(pivot)
```

```python
pivot = pd.pivot_table(
    data=df,
    index='Region',
    values='Sales',
    sort=True
)

print(pivot)
```
