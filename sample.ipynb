```python
import pandas as pd

# Sample DataFrame
data = {
    'Department': ['IT', 'IT', 'HR', 'HR', 'Finance', 'Finance'],
    'Salary': [60000, 75000, 50000, 52000, 80000, 82000],
    'YearsExperience': [2, 3, 5, 4, 6, 7]
}

df = pd.DataFrame(data)
```

```python
df.groupby('Department')['Salary'].mean()
```

```python
# Custom multiple aggregation
df.groupby('Department').agg({
    'Salary': ['mean', 'sum'],
    'YearsExperience': 'max'
})

```

```python
df.groupby('Department').agg(
    lambda x: x.max() - x.min()
)
```

```python
df.groupby('Department')['Salary'].apply(lambda x: (x.max() + x.min()) / 2)
```

```python
df.groupby('Department').agg(
    avg_salary=('Salary', 'mean'),
    total_exp=('YearsExperience', 'sum')
)
```

```python
df.groupby('Department').agg(
    avg_salary=('Salary', 'mean'),
    total_exp=('YearsExperience', 'sum')).reset_index()
```

```python
df.groupby('Department')['Salary'].sum().reset_index()
```

```python
df.describe()                     # Summary for entire DataFrame
```

```python
df.groupby('Department').describe()  # Summary per group
```

```python
df.groupby('Department').describe().reset_index()
```

```python
# GroupBy is faster when the column is category type.
df['Department'] = df['Department'].astype('category')

```

```python
# Use sort=False in GroupBy
df.groupby('Department', sort=False).agg({...})

# Pre-filter Columns for Aggregation
df[['Department', 'Salary', 'YearsExperience']].groupby('Department').agg({...})
```

```python
# Use Vectorized Aggregation Instead of Loops
# Slow
df.groupby('Department').apply(lambda x: x['Salary'].max() - x['Salary'].min())
```

```python
# Fast
df.groupby('Department')['Salary'].agg(lambda x: x.max() - x.min())
```

```python
# Use Named Aggregation 
df.groupby('Department').agg(
    avg_salary=('Salary', 'mean'),
    max_exp=('YearsExperience', 'max')
)
```

```python
# Defensive Aggregation 
safe_cols = ['Salary', 'YearsExperience']
available_cols = [col for col in safe_cols if col in df.columns]

df.groupby('Department')[available_cols].agg('mean')
```

```python
# ✅ 7. Drop NA Before Grouping If Needed
df = df.dropna(subset=['Department'])
```

```python
# Use .values if only one column is needed after aggregation
grouped = df.groupby('Department')['Salary'].sum().values
grouped
```

```python
# ✅ 9. Parallel Processing with Dask (for VERY large data)
import dask.dataframe as dd

ddf = dd.from_pandas(df, npartitions=4)
ddf.groupby('Department')['Salary'].mean().compute()
```

```python
import pandas as pd

df = pd.DataFrame({
    'date': ['2023-01-01', '2023-01-01', '2023-01-02'],
    'category': ['A', 'B', 'A'],
    'value': [10, 20, 30]
})

pivot_df = pd.pivot_table(
    df,
    index='date',          # You want to group by date
    columns='category',    # You want categories to become columns
    values='value',
    aggfunc='sum',
    fill_value=0           # Optional: fill NaNs with 0
)

# Flatten multi-index columns (optional if you use multiple aggregations)
pivot_df.columns.name = None

# 🔑 Step: Reset index so 'date' becomes a column again
pivot_df = pivot_df.reset_index()
```
